```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)

print("DSP Laboratory setup successful!")
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)  # Unit impulse
step = (n >= 0).astype(float)   # Unit step
ramp = n * step                 # Unit ramp
expo = 0.8**n * step            # Exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 21)
a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = a**n
    ax.stem(n, x)
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

plt.figure(figsize=(12, 4))

plt.plot(n, x1, 'o-', label='cos(2πn/8)')
plt.plot(n, x2, 's--', label='cos(2πn/7.5)')

plt.xlabel("n")
plt.ylabel("Amplitude")
plt.title("Comparison of Discrete-Time Cosine Signals")
plt.legend()
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio, display

# Read WAV audio
fs, x = wavfile.read("/content/my_voice.wav")

# Normalize audio
x = x.astype(float)
x = x / max(np.max(np.abs(x)), 1)

# Calculate time
t = np.arange(len(x)) / fs

# Display information
print("Sampling Rate:", fs, "Hz")
print("Duration:", round(len(x)/fs, 2), "seconds")
print("Number of Samples:", len(x))

# Plot waveform
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("Task 1.3 - My Voice Recording")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# Play audio
display(Audio(x, rate=fs))
```
